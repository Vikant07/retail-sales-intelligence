#  4. Product Analysis

This section evaluates product performance by identifying the best-selling products, most profitable products, loss-making products, and product category trends to support inventory and business decisions.

```python

def profitability_summary(dimension):
    result = (
        df.groupby(dimension, dropna=False)
          .agg(
              Total_Sales=("Sales", "sum"),
              Total_Profit=("Profit", "sum"),
              Distinct_Orders=("Order_ID", "nunique")
          )
          .reset_index()
    )

    result["Profit_Margin_Pct"] = np.where(
        result["Total_Sales"] != 0,
        result["Total_Profit"] / result["Total_Sales"] * 100,
        np.nan
    )

    return result.sort_values(
        "Total_Profit", ascending=True
    ).round(2)


for dimension in [
    "Category",
    "Sub-Category",
    "Region",
    "Market",
    "Segment"
]:
    print(f"\nProfitability by {dimension}")
    display(profitability_summary(dimension))
```

### Shipping Mode Analysis

```python

shipping_analysis = (
    df.groupby("Ship_Mode", dropna=False)
      .agg(
          Distinct_Orders=("Order_ID", "nunique"),
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Shipping_Cost=("Shipping_Cost", "sum")
      )
      .reset_index()
)

shipping_analysis["Avg_Shipping_Days"] = (
    df.assign(
        Shipping_Days=(
            df["Ship_Date"] - df["Order_Date"]
        ).dt.days
    )
    .groupby("Ship_Mode")["Shipping_Days"]
    .mean()
    .reindex(shipping_analysis["Ship_Mode"])
    .to_numpy()
)

display(shipping_analysis.round(2))
```
